```python
df_drop = df.dropna()

print("\n========== Shape After Removing Missing Rows ==========")
print(df_drop.shape)
```

```python
df_mean = df.copy()
```

```python
df_mean["Age"] = df_mean["Age"].fillna(df_mean["Age"].mean())

print("\n========== Missing Values After Mean ==========")
print(df_mean["Age"].isnull().sum())
```

```python
df_median = df.copy()
```

```python
df_median["Age"] = df_median["Age"].fillna(df_median["Age"].median())

print("\n========== Missing Values After Median ==========")
print(df_median["Age"].isnull().sum())
```

```python
df_mode = df.copy()
```

```python
df_mode["Embarked"] = df_mode["Embarked"].fillna(df_mode["Embarked"].mode()[0])

print("\n========== Missing Values After Mode ==========")
print(df_mode["Embarked"].isnull().sum())
```

```python
print("\n========== Missing Values (Mean Method) ==========")
print(df_mean.isnull().sum())

print("\n========== Missing Values (Median Method) ==========")
print(df_median.isnull().sum())

print("\n========== Missing Values (Mode Method) ==========")
print(df_mode.isnull().sum())


# ============================================================
# IDENTIFY FEATURE VARIABLE AND TARGET VARIABLE
# ============================================================

# Target Variable (Dependent Variable)

Y = df["Survived"]

# Feature Variables (Independent Variables)

X = df.drop("Survived", axis=1)

print("\n========== Feature Variables ==========")
print(X.head())

print("\n========== Target Variable ==========")
print(Y.head())


# ============================================================
# Display Number of Rows and Columns Separately
# ============================================================

print("\n========== Dataset Size ==========")

print("Number of Rows :", df.shape[0])

print("Number of Columns :", df.shape[1])


# ============================================================
# Display Numerical Columns
# ============================================================

print("\n========== Numerical Columns ==========")

print(df.select_dtypes(include=["int64", "float64"]).columns)


# ============================================================
# Display Categorical Columns
# ============================================================

print("\n========== Categorical Columns ==========")

print(df.select_dtypes(include=["object"]).columns)
```
